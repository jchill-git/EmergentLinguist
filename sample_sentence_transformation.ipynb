```python
#transform sample sentences
with open("sentences.txt", "r", encoding="utf-8") as file:
    sentences = file.read().splitlines()
transformed_sentences = [sentence_transform(sent) for sent in sentences]
```

```python
print(transformed_sentences)
```

```python
# English and transformed sentence side by side, to check each rule
for en, cl in zip(sentences, transformed_sentences):
    print(f"{en}\n  -> {cl}")
```

```python
with open("transformed_sentences.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(transformed_sentences))
```
